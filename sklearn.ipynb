```python
import sklearn as sk
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from sklearn.linear_model import LinearRegression
```

```python
x=np.array([[1,2],[3,4],[5,6],[7,8]])
y=np.dot(x,np.array([1,2]))+3
reg=LinearRegression().fit(x,y)
print(reg.score(x,y))
print(reg.coef_)
print(reg.intercept_)
print(reg.predict(np.array([[9,10]])))
reg.get_params()
```

```python
german=df2.drop(columns=['Saving accounts', 'Checking account','Purpose'],inplace=True)
german=df2.dropna()
print(german.shape)
german['Age']=german['Age'].fillna(german['Age'].mean)
german.info()
```

```python
corr=german.corr()
print(corr[(corr)>0.1])
```

```python
x=german.drop(columns=['Risk'])
y=german['Risk']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=32)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
print(y_pred)
y_predcounts=pd.Series(y_pred).value_counts()
print(y_test)
print(y_predcounts)
pd.Series(y_test).value_counts()
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
print(confusion_matrix(y_test,y_pred))
```

```python
print(accuracy_score(y_test,y_pred))
```

```python
print(classification_report(y_test,y_pred))
```
