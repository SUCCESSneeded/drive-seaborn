```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
penguins = sns.load_dataset("penguins")
print(penguins)
sns.histplot(data=penguins, x="flipper_length_mm")
```

```python
sns.histplot(data=penguins, y="flipper_length_mm")
```

```python
sns.histplot(data=penguins, x="flipper_length_mm",binwidth=6) # decides the width of bar on x-axis
```

```python
sns.histplot(data=penguins, x="flipper_length_mm", bins=10) #decide number of bars
```

```python
#with kde
sns.histplot(data=penguins, x="flipper_length_mm", kde=True)
```

```python
#If neither x nor y is assigned, the dataset is treated as wide-form, and a histogram is drawn for each numeric column:
sns.histplot(data=penguins)
```

```python
sns.histplot(data=penguins, x="flipper_length_mm", hue="species", multiple="stack")
```

```python
sns.histplot(penguins, x="flipper_length_mm", hue="species", element="step")
#element->appearance of the histogram ->bars,steps,poly
```

```python
sns.histplot(
    penguins, x="bill_length_mm", hue="island", element="step",
    stat="count", common_norm=False,
)
#stats what y axis represents -> count,frequency,density,probability
#common_norm -> suggest if normalization will happen together or individual
```

```python
tips = sns.load_dataset("tips")
sns.histplot(data=tips, x="size", stat="percent", discrete=True)
```

```python
sns.histplot(data=tips, x="day", shrink=1)
```

```python
sns.histplot(data=tips, x="day", hue="sex", multiple="dodge", shrink=.8)
```

```python
planets=sns.load_dataset("planets")
planets.head()
```

```python
sns.histplot(data=planets,x="distance")
```

```python
sns.histplot(data=pd.DataFrame(planets),x="distance",log_scale=True)
```

```python
sns.histplot(data=planets, x="distance", log_scale=True, fill=False)
```

```python
sns.histplot(data=planets, x="distance", log_scale=True, element="step", fill=False)
```

```python
sns.histplot(
    data=planets, x="distance", hue="method",
    hue_order=["Radial Velocity", "Transit"],
    log_scale=True, element="step", fill=False,
    cumulative=True, stat="density", common_norm=False,
)   #commulative means adding all previous value to current and ploting
```

```python
sns.histplot(
    data=planets, x="distance", hue="method",
    hue_order=["Radial Velocity", "Transit"],
    log_scale=True, element="step", fill=False,
    cumulative=False, stat="density", common_norm=False,
)
```

```python
#When both x and y are assigned, a bivariate histogram is computed and shown as a heatmap:
sns.histplot(penguins, x="bill_depth_mm", y="body_mass_g")
```

```python
sns.histplot(penguins, x="bill_depth_mm", y="body_mass_g",hue="species")
```

```python
sns.histplot(
    planets, x="year", y="distance",
    bins=30, discrete=(True, False), log_scale=(False, True),
)
```

```python
sns.histplot(
    planets, x="year", y="distance",
    bins=30, discrete=(True, False), log_scale=(False, True),thresh=None
)
#higher thresh fewer visible cells.Gives condition like if thresh=3 then bin should contains atleast 3 obs
```

```python
sns.histplot(
    planets, x="year", y="distance",
    bins=30, discrete=(True, False), log_scale=(False, True),
    pthresh=.05, pmax=.9,
)
#if pthresh->0.01 hides the cell whose proportion is less than 1%
#pmax->changes how colors are scaled
```

```python
sns.histplot(
    planets, x="year", y="distance",
    bins=30, discrete=(True, False), log_scale=(False, True),
    cbar=True, cbar_kws=dict(shrink=0.85),
)
#cbar_kws->dict is pas containing the labels to the colorbar
```
