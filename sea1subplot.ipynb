```python
#dataframe.eval()-> MODIFY THE COLUMN OF DATAFRAME

tips1=pd.DataFrame(tips)
tip = tips1.eval("tip / total_bill").rename("tip_rate")
sns.scatterplot(data=tips, x="total_bill", y="tip", hue=tip)
```

```python
#Control the range of marker areas with sizes, and set legend="full" to force every unique value to appear in the legend:

sns.scatterplot(
    data=tips, x="total_bill", y="tip", hue="size", size="size",
    sizes=(20, 200), legend="full"
)
```

```python
#hue_norm to control the quantitative hue mapping:

sns.scatterplot(
    data=tips, x="total_bill", y="tip", hue="size", size="size",
    sizes=(20, 200), hue_norm=(0, 6), legend="full"
)
```

```python
markers = {"lunch": "s", "Dinner": "X"}
sns.scatterplot(data=tips, x="total_bill", y="tip", style="time",hue='size',hue_norm=(0,7),size='size',legend='auto',sizes=(0,200), markers=markers)
```
