# **scatter plot**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

tips = {
    "total_bill": [16.99, 10.34, 21.01, 23.68, 24.59, 25.29, 8.77, 26.88, 15.04, 14.78, 10.27, 35.26, 15.42, 18.43, 14.83, 21.58, 10.33, 16.29, 16.97, 20.65],
    "smoker": ["No", "No", "No", "Yes", "No", "No", "No", "No", "Yes", "No", "No", "No", "No", "No", "Yes", "No", "No", "No", "No", "No"],
    "sex": ["Female", "Male", "Male", "Male", "Female", "Male", "Male", "Male", "Male", "Male", "Male", "Female", "Male", "Male", "Female", "Male", "Female", "Male", "Female", "Male"],
    "day": ["Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun", "Sun"],
    "time": ["Dinner", "Dinner", "lunch", "Dinner", "Dinner", "lunch", "Dinner", "lunch", "Dinner", "lunch", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner"],
    "size": [2, 3, 3, 2, 4, 4, 2, 4, 2, 2, 2, 4, 2, 4, 2, 2, 5, 3, 3, 3],
    "tip": [1.01, 1.66, 3.50, 3.31, 3.61, 4.71, 2.00, 3.12, 1.96, 3.23, 1.71, 5.00, 1.57, 3.00, 3.02, 3.92, 1.67, 3.51, 3.50, 3.35]
}

# Create the DataFrame
df = pd.DataFrame(tips)

# Display the first few rows
print(df.head())
```

```python
#axes level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#relplot -> gives same graph as scatter plot
#figure level

sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
```

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
#seaborn.scatterplot(data=None, *, x=None, y=None, hue=None, size=None, style=None,
#palette=None, hue_order=None, hue_norm=None, sizes=None, size_order=None, size_norm=None, markers=True, style_order=None, legend='auto', ax=None, **kwargs)
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="size",style='size')
```

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="size", palette='colorblind')
#palette-> deep,dark,"muted", "bright", "pastel",colorblind
```

```python
#dataframe.eval()-> MODIFY THE COLUMN OF DATAFRAME

tips1=pd.DataFrame(tips)
tip = tips1.eval("tip / total_bill").rename("tip_rate")
sns.scatterplot(data=tips, x="total_bill", y="tip", hue=tip)
```

```python
#Control the range of marker areas with sizes, and set legend="full" to force every unique value to appear in the legend:

sns.scatterplot(
    data=tips, x="total_bill", y="tip", hue="size", size="size",
    sizes=(20, 200), legend="full"
)
```

```python
#hue_norm to control the quantitative hue mapping:

sns.scatterplot(
    data=tips, x="total_bill", y="tip", hue="size", size="size",
    sizes=(20, 200), hue_norm=(0, 6), legend="full"
)
```

```python
markers = {"lunch": "s", "Dinner": "X"}
sns.scatterplot(data=tips, x="total_bill", y="tip", style="time",hue='size',hue_norm=(0,7),size='size',legend='auto',sizes=(0,200), markers=markers)
```

# Additional keyword arguments

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", s=100, color="red", marker="+")
```

```python
index = pd.date_range("1 1 2000", periods=100, freq="m", name="date")  #daterange->parameters ->(start,end,freq),(start,periods='',freq)
data = np.random.randn(100, 4).cumsum(axis=0)  #cumsum keeps on adding previous values to the current value
wide_df = pd.DataFrame(data, index, ["a", "b", "c", "d"])
sns.scatterplot(data=wide_df)
```
