# **KDE PLOT**

```python
'''kdeplot(data=None, *, x=None, y=None, hue=None, weights=None, palette=None, hue_order=None, hue_norm=None, color=None,
 fill=None, multiple='layer', common_norm=True, common_grid=False, cumulative=False, bw_method='scott', bw_adjust=1,
  warn_singular=True, log_scale=None, levels=10, thresh=0.05, gridsize=200, cut=3, clip=None, legend=True, cbar=False,
   cbar_ax=None, cbar_kws=None, ax=None, **kwargs)'''
```

```python
'''A kernel density estimate (KDE) plot is a method for visualizing the distribution of observations in a dataset, analogous to a histogram.
 KDE represents the data using a continuous probability density curve in one or more dimensions.'''
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
tips = sns.load_dataset("tips")
sns.kdeplot(data=tips, x="total_bill")
```

```python
iris = sns.load_dataset("iris")
sns.kdeplot(data=iris)
```

```python
#bw_adjust -> bandwidth of kde
#bw_adjust -> =1 default smoothing
#bw_adjust -> <1 less smooothing
#bw_adjust -> >1 more smoothing
sns.kdeplot(data=tips, x="total_bill", bw_adjust=.2)
```

```python
ax= sns.kdeplot(data=tips, x="total_bill", bw_adjust=5, cut=0) # extension of values default =2
```

```python
sns.kdeplot(data=tips, x="total_bill", hue="time")
```

```python
sns.kdeplot(data=tips, x="total_bill", hue="time", multiple="stack")
```

```python
sns.kdeplot(data=tips, x="total_bill", hue="time", multiple="fill")
```

```python
sns.kdeplot(
    data=tips, x="total_bill", hue="time",
    cumulative=True, common_norm=False, common_grid=True,
)
```

```python
tips_agg = (tips
    .groupby("size")
    .agg(total_bill=("total_bill", "mean"), n=("total_bill", "count"))
)
sns.kdeplot(data=tips_agg, x="total_bill", weights="n")
```

```python
sns.kdeplot(
   data=tips, x="total_bill", hue="size",
   fill=True, common_norm=False, palette="winter",
   alpha=.5, linewidth=0,
)
```

```python
geyser = sns.load_dataset("geyser")
sns.kdeplot(data=geyser, x="waiting", y="duration")
```

```python
sns.kdeplot(data=tips, x="total_bill", y="size")
```

```python
sns.kdeplot(data=geyser, x="waiting", y="duration", hue="kind")
```

```python
sns.kdeplot(
    data=geyser, x="waiting", y="duration", hue="kind", fill=True
)
```

```python
#Show fewer contour levels, covering less of the distribution:
#thresh -> display regions whose density is below this threshold

sns.kdeplot(
    data=geyser, x="waiting", y="duration", hue="kind",
    levels=5, thresh=.1,
)
```

```python
sns.kdeplot(
    data=geyser, x="waiting", y="duration",
    fill=True, thresh=0, levels=100, cmap="mako",
)
```
